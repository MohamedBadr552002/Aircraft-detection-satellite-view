```python
from google.colab import userdata
API_Key = userdata.get('API_Key')

from roboflow import Roboflow
rf = Roboflow(api_key=API_Key)

# # List all available workspaces
# workspaces = rf.workspace()
# print(workspaces)



project = rf.workspace("satobjectdetection").project("fighter-jet-yolo-od-zbwpt")
version = project.version(1)
dataset = version.download("yolov11")
```

```python
DATASET_PATH ="/content/fighter-jet-yolo-OD-1"
DATASET_YML_FILE = "/content/fighter-jet-yolo-OD-1/data.yaml"
```

## YOLO 11 Fine-tuning

```python
# Train yolo11 on custom dataset
from ultralytics import YOLO
# Load a model
model = YOLO('yolo11n.pt')  # or yolo11s.pt, yolo11m.pt, yolo11l.pt, yolo11x.pt

# Train with early stopping
results = model.train(
    data=DATASET_YML_FILE,
    epochs=100,
    patience=10,  # Early stopping patience (epochs to wait without improvement)
    batch=16,
    imgsz=512,
    save=True,
    device=0  # Use GPU if available
)
```

```python
results = model.val()
```
